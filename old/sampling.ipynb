```python
import os
import random
import shutil

# ---- À MODIFIER ----
source_folder = "C:/Users/jean-/Desktop/Cours ENSAE/3A ENSAE/NLP/Projet final/NLP_ENSAE_Project/legislatives"
to_avoid_folder = "C:/Users/jean-/Desktop/Cours ENSAE/3A ENSAE/NLP/Projet final/NLP_ENSAE_Project/annotations Julien"
output_folder = "C:/Users/jean-/Desktop/Cours ENSAE/3A ENSAE/NLP/Projet final/NLP_ENSAE_Project/annotations JL"
n = 300
# --------------------

# 1. Récupérer la liste des fichiers à exclure sous forme de set (pour la rapidité de recherche)
try:
    files_to_avoid = set(os.listdir(to_avoid_folder))
except FileNotFoundError:
    # Si le dossier n'existe pas, aucun fichier n'est à éviter
    files_to_avoid = set() 

# 2. Lister les fichiers .txt du dossier source qui NE SONT PAS dans le dossier à éviter
valid_files = [
    f for f in os.listdir(source_folder) 
    if f.endswith(".txt") and f not in files_to_avoid
]

# 3. Sample aléatoire reproductible sur la liste filtrée
random.seed(42)
sample = random.sample(valid_files, min(n, len(valid_files)))

# 4. Création du dossier cible et copie
os.makedirs(output_folder, exist_ok=True)
for f in sample:
    shutil.copy(os.path.join(source_folder, f), os.path.join(output_folder, f))

```
